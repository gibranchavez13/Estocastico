```python
import numpy as np
from scipy.stats import norm
```

## Datos iniciales

```python
r0 = 0.0654
t = 1/12
limite_inferior = r0
limite_superior = r0 + 0.01
```

```python
print("Intervalo de probabilidad:")
print(f"De {limite_inferior:.4f} a {limite_superior:.4f}")
print(f"De {limite_inferior*100:.2f}% a {limite_superior*100:.2f}%")
```

### Black

```python
alpha_black = -0.0011939577039274916
sigma_black = 0.010097927468342677
```

```python
media_black = r0 + alpha_black*t
var_black = sigma_black**2 * t
desv_black = np.sqrt(var_black)

prob_black = norm.cdf((limite_superior - media_black) / desv_black) - norm.cdf((limite_inferior - media_black) / desv_black)
```

```python
print("Modelo Black")
print(f"Esperanza: {media_black:.8f} = {media_black*100:.4f}%")
print(f"Varianza: {var_black:.12f}")
print(f"Desviación estándar: {desv_black:.8f} = {desv_black*100:.4f}%")
print(f"Probabilidad: {prob_black:.6f} = {prob_black*100:.2f}%")
```

### Rendleman-Bartter

```python
mu_rb = -0.01739368746141955
sigma_rb = 0.002594712378987028
```

```python
media_rb = r0 * np.exp(mu_rb*t)
var_rb = r0**2 * np.exp(2*mu_rb*t) * (np.exp(sigma_rb**2*t) - 1)
desv_rb = np.sqrt(var_rb)

# Parámetros de la normal asociada al logaritmo
media_log_rb = np.log(r0) + (mu_rb - 0.5*sigma_rb**2)*t
desv_log_rb = sigma_rb * np.sqrt(t)

prob_rb = norm.cdf((np.log(limite_superior) - media_log_rb) / desv_log_rb) - norm.cdf((np.log(limite_inferior) - media_log_rb) / desv_log_rb)
```

```python
print("Modelo Rendleman-Bartter")
print(f"Esperanza: {media_rb:.8f} = {media_rb*100:.4f}%")
print(f"Varianza: {var_rb:.12f}")
print(f"Desviación estándar: {desv_rb:.8f} = {desv_rb*100:.4f}%")
print(f"Probabilidad: {prob_rb:.6f} = {prob_rb*100:.2f}%")
```

### Vasicek

```python
a_vas = 0.06137406553078901
b_vas = 0.06013623432065643
sigma_vas = 0.010082416954693387
```

```python
media_vas = r0*np.exp(-a_vas*t) + b_vas*(1 - np.exp(-a_vas*t))
var_vas = (sigma_vas**2 / (2*a_vas)) * (1 - np.exp(-2*a_vas*t))
desv_vas = np.sqrt(var_vas)

prob_vas = norm.cdf((limite_superior - media_vas) / desv_vas) - norm.cdf((limite_inferior - media_vas) / desv_vas)
```

```python
print("Modelo Vasicek")
print(f"Esperanza: {media_vas:.8f} = {media_vas*100:.4f}%")
print(f"Varianza: {var_vas:.12f}")
print(f"Desviación estándar: {desv_vas:.8f} = {desv_vas*100:.4f}%")
print(f"Probabilidad: {prob_vas:.6f} = {prob_vas*100:.2f}%")
```
