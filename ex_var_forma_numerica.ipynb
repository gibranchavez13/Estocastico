# Resolución de forma numérica (E[$r_t$] y Var($r_t$))

# Datos inciales

```python
import numpy as np
```

```python
# Condición inicial
r0 = 0.0654

# Horizonte: un mes
T = 1/12

# Número de pasos dentro del mes
N = 30
dt = T / N

# Número de simulaciones
M = 10000

# Semilla para reproducibilidad
np.random.seed(42)
```

# Modelo de Black
## $dr_t = \alpha dt + \sigma dW_t$ 

```python
# Parámetros Black
alpha = -0.0011939577039274916
sigma = 0.010097927468342677

# Matriz de simulaciones
r_black = np.zeros((M, N + 1))
r_black[:, 0] = r0

# Simulación Euler-Maruyama
for i in range(N):
    Z = np.random.normal(0, 1, M)
    dW = np.sqrt(dt) * Z
    
    r_black[:, i + 1] = r_black[:, i] + alpha * dt + sigma * dW

# Valores finales al mes
r_final_black = r_black[:, -1]

# Esperanza y varianza numéricas
esperanza_black_num = np.mean(r_final_black)
varianza_black_num = np.var(r_final_black, ddof=1)
desviacion_black_num = np.sqrt(varianza_black_num)

print("Modelo Black - Forma numérica")
print(f"Esperanza numérica a un mes: {esperanza_black_num:.8f} = {esperanza_black_num*100:.4f}%")
print(f"Varianza numérica a un mes: {varianza_black_num:.10f}")
print(f"Desviación estándar numérica: {desviacion_black_num:.8f} = {desviacion_black_num*100:.4f}%")
```

# Modelo de Rendleman-Batter
## $dr_t = \mu r_t dt + \sigma r_t dW_t$ 

```python
# Parámetros Rendleman-Bartter
mu = -0.01739368746141955
sigma = 0.002594712378987028

# Matriz de simulaciones
r_rb = np.zeros((M, N + 1))
r_rb[:, 0] = r0

# Simulación Euler-Maruyama
for i in range(N):
    Z = np.random.normal(0, 1, M)
    dW = np.sqrt(dt) * Z
    
    r_rb[:, i + 1] = r_rb[:, i] + mu * r_rb[:, i] * dt + sigma * r_rb[:, i] * dW

# Valores finales al mes
r_final_rb = r_rb[:, -1]

# Esperanza y varianza numéricas
esperanza_rb_num = np.mean(r_final_rb)
varianza_rb_num = np.var(r_final_rb, ddof=1)
desviacion_rb_num = np.sqrt(varianza_rb_num)

print("Modelo Rendleman-Bartter - Forma numérica")
print(f"Esperanza numérica a un mes: {esperanza_rb_num:.8f} = {esperanza_rb_num*100:.4f}%")
print(f"Varianza numérica a un mes: {varianza_rb_num:.12f}")
print(f"Desviación estándar numérica: {desviacion_rb_num:.8f} = {desviacion_rb_num*100:.4f}%")
```

# Modelo de Vasicek
## $dr_t = a(b - r_t)dt + \sigma dW_t$ 

```python
# Parámetros Vasicek
a = 0.06137406553078901
b = 0.06013623432065643
sigma = 0.010082416954693387

# Matriz de simulaciones
r_vasicek = np.zeros((M, N + 1))
r_vasicek[:, 0] = r0

# Simulación Euler-Maruyama
for i in range(N):
    Z = np.random.normal(0, 1, M)
    dW = np.sqrt(dt) * Z
    
    r_vasicek[:, i + 1] = r_vasicek[:, i] + a * (b - r_vasicek[:, i]) * dt + sigma * dW

# Valores finales al mes
r_final_vasicek = r_vasicek[:, -1]

# Esperanza y varianza numéricas
esperanza_vasicek_num = np.mean(r_final_vasicek)
varianza_vasicek_num = np.var(r_final_vasicek, ddof=1)
desviacion_vasicek_num = np.sqrt(varianza_vasicek_num)

print("Modelo Vasicek - Forma numérica")
print(f"Esperanza numérica a un mes: {esperanza_vasicek_num:.8f} = {esperanza_vasicek_num*100:.4f}%")
print(f"Varianza numérica a un mes: {varianza_vasicek_num:.10f}")
print(f"Desviación estándar numérica: {desviacion_vasicek_num:.8f} = {desviacion_vasicek_num*100:.4f}%")
```
